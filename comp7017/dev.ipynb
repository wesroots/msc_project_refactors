```python
from comp7017.data import load_classes, prepare_xy_splits, prepare_ttv_splits
from comp7017.edge_cnn import _build_m4
from comp7017.kernels import generate_prior_kernels, generate_random_kernels, construct_initializer

from keras import layers
```

```python
X, y = prepare_xy_splits()

ttv_splits = prepare_ttv_splits(X, y)
```

```python
prior_kernels = generate_prior_kernels()
glorot_kernels = generate_random_kernels(shape=(3,3,1,14))
initializer = construct_initializer(prior_kernels, glorot_kernels)
```

```python
normalizer = layers.Normalization(axis=None)
normalizer.adapt(ttv_splits["X_train"])
```

```python
_build_m4(ttv_splits["X_train"], initializer=initializer, normalizer=normalizer)
```
